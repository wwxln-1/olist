## 1.1 — Data Analytics: Definition & Role in Business  <span style="background:#1e3a5f;color:white;padding:2px 10px;border-radius:12px;font-size:12px;font-weight:600;">P1</span>

**Data analytics** is the systematic process of examining, cleaning, transforming, and modelling raw data to discover useful information, draw conclusions, and support decision-making in organisations (McKinney, 2022). It converts large volumes of operational records into patterns, trends, and actionable intelligence that management can act upon.

In a business context, data analytics plays three interconnected roles:
1. **Understanding performance** — describing what has already happened (descriptive analytics).
2. **Forecasting outcomes** — predicting what is likely to happen next (predictive analytics).
3. **Recommending action** — prescribing the optimal course of action (prescriptive analytics).

Marr (2017) argues that organisations failing to utilise their data are operating at a strategic disadvantage. For Olist — a Brazilian e-commerce marketplace — applying data analytics to 112,650 transaction records enables identification of top-performing product categories, revenue trends, and demand patterns that would otherwise remain invisible.



## 1.2 — Key Statistical Terms with Dataset Examples  <span style="background:#1e3a5f;color:white;padding:2px 10px;border-radius:12px;font-size:12px;font-weight:600;">P1</span>

### 📌 Population vs. Sample

| Concept | Definition | Olist Dataset Example |
|---------|------------|----------------------|
| **Population** | The complete set of all observations of interest (Bruce & Bruce, 2020). | All 112,650 order-item records in the merged `main` DataFrame — every transaction on the platform. |
| **Sample** | A subset drawn from the population for analysis when examining the full population is impractical. | A 10% random sample of **11,265 records** drawn using `main.sample(frac=0.10, random_state=42)`. |

---

### 📌 Categorical Data

| Type | Definition | Olist Dataset Example |
|------|------------|----------------------|
| **Nominal** | Categories with no inherent order — labels or names only (Provost & Fawcett, 2013). | `customer_state` (SP, RJ, MG…) and `order_status` (delivered, cancelled…) — no value is "greater" than another. |
| **Ordinal** | Categories with a meaningful rank order, but unequal intervals between ranks. | Review scores (1–5 stars in `olist_order_reviews_dataset.csv`) — 5 > 3, but the gap between 4 and 5 may not equal the gap between 1 and 2. |

---

### 📌 Numerical Data

| Type | Definition | Olist Dataset Example |
|------|------------|----------------------|
| **Continuous** | Can take any value within a range, including decimals — measurements (Bruce & Bruce, 2020). | `price` (R\$58.90, R\$239.93), `freight_value`, and the derived `revenue` column. |
| **Discrete** | Only specific countable values; gaps exist between possible values. | `order_item_id` (1, 2, 3…) and `product_photos_qty` — an order cannot contain 2.7 items. |

---

### 📌 Descriptive vs. Predictive Analytics

| Type | Question Answered | Olist Example |
|------|------------------|---------------|
| **Descriptive** | *What happened?* — summarises historical data using statistics and visualisations. | Computing mean monthly revenue and top product categories from 2016–2018 transactions. |
| **Predictive** | *What will happen?* — uses models to forecast future outcomes from historical patterns (James et al., 2021). | Using linear regression to estimate revenue from price; using moving average to project next-month revenue. |

## 1.6 — The Three Types of Analytics and Their Industry Use  <span style="background:#2563eb;color:white;padding:2px 10px;border-radius:12px;font-size:12px;font-weight:600;">M1</span>

### 1. Descriptive Analytics — *"What happened?"*
Descriptive analytics uses historical data, aggregation, and visualisation to summarise past events (Provost & Fawcett, 2013).

**Industry Example:** Amazon analyses daily sales volumes, return rates, and customer ratings per product line. Operational managers use dashboards built on descriptive analytics to identify declining categories before they materially impact revenue.

**In this examination:** Monthly revenue totals, top-10 product category frequencies, and revenue distribution histograms were produced to describe Olist's historical transaction patterns from 2016–2018.

---

### 2. Predictive Analytics — *"What will happen?"*
Predictive analytics applies statistical algorithms and machine learning to historical patterns to forecast future outcomes (James et al., 2021).

**Industry Example:** Netflix uses predictive models to forecast viewer demand for content genres by region and season, informing content acquisition budgets and release scheduling. Streaming rights for a series are purchased months in advance based on predicted audience size.

**In this examination:** Simple Linear Regression (R² = 0.994) and a 3-Month Rolling Moving Average were applied to Olist transaction data to forecast revenue and quantify the price–revenue relationship.

---

### 3. Prescriptive Analytics — *"What should we do?"*
Prescriptive analytics recommends optimal actions to achieve desired outcomes, combining predictive outputs with optimisation methods (Marr, 2017).

**Industry Example:** UPS's ORION system processes millions of delivery permutations daily, prescribing the most fuel-efficient driver routes. This prescriptive system saves over 100 million miles and 10 million gallons of fuel annually.

**In this examination:** Prescriptive analytics was not formally applied, but the foundation exists: moving average revenue forecasts could feed into a prescriptive inventory-replenishment model that automatically recommends stock orders for high-demand categories (e.g., `cama_mesa_banho`) ahead of seasonal peaks.

The Interquartile Range (IQR) method is a robust technique for detecting outliers that is resistant to the influence of extreme values (Bruce & Bruce, 2020). Values below Q1 − 1.5×IQR or above Q3 + 1.5×IQR are flagged as statistical outliers.


```python
Q1 = main["revenue"].quantile(0.25)
Q3 = main["revenue"].quantile(0.75)

IQR = Q3 - Q1

print(Q1,Q3,IQR)
```

```python
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = main[
    (main["revenue"] < lower) |
    (main["revenue"] > upper)
]

print(len(outliers))
```

## 3.1 — What is Predictive Analytics?  `P5`

**Predictive analytics** applies statistical algorithms and machine learning to historical data patterns to forecast future outcomes or estimate unknown values (James et al., 2021). Unlike descriptive analytics — which summarises *what happened* — predictive analytics answers *what is likely to happen next*.

The two techniques applied in this examination are:

| Technique | Type | Business Question Answered |
|-----------|------|---------------------------|
| **Simple Linear Regression** | Structural relationship | How does product price drive total revenue per item? |
| **Multiple Linear Regression** | Multi-variable structure | How do price AND freight together predict revenue? |
| **3-Month Rolling Moving Average** | Temporal forecasting | What will next month's revenue approximately be? |
| **Exponential Smoothing** | Temporal forecasting | What is the weighted trend forecast giving more weight to recent months? |

Both regression and time-series methods are applied to provide a comprehensive predictive view of the Olist dataset.



## 3.2 — Simple Linear Regression: Price → Revenue  `P5` `P6`

Simple Linear Regression models the linear relationship between one predictor (price) and one response variable (revenue).  
**Model form:** Revenue = β₀ + β₁ × Price + ε


```python
from sklearn.linear_model import LinearRegression

X = main[["price"]]
y = main["revenue"]

model = LinearRegression()
model.fit(X,y)

print(model.coef_)
print(model.intercept_)
```

```python
from sklearn.metrics import r2_score

pred = model.predict(X)

print(r2_score(y,pred))
```


## 3.3 — Multiple Linear Regression: Price + Freight → Revenue  `P5` `P6`

Multiple Linear Regression extends the simple model by adding `freight_value` as a second predictor, capturing the additive contribution of both components to total revenue.  
**Model form:** Revenue = β₀ + β₁×Price + β₂×Freight + ε


```python
# ─── 3.3 Multiple Linear Regression ──────────────────────────────────────────
X_multi = main[["price", "freight_value"]]

mlr = LinearRegression()
mlr.fit(X_multi, y)
y_pred_mlr = mlr.predict(X_multi)


mae_mlr = mean_absolute_error(y, y_pred_mlr)
rmse_mlr = np.sqrt(mean_squared_error(y, y_pred_mlr))

print(f"  Intercept (β₀)         : {mlr.intercept_:.8f}")
print(f"  Coef — price  (β₁)     : {mlr.coef_[0]:.8f}")
print(f"  Coef — freight (β₂)    : {mlr.coef_[1]:.8f}")
print(f"  R² (Goodness-of-fit)   : {r2_mlr:.8f}")
print(f"  MAE                    : R${mae_mlr:.8f}")
print(f"  RMSE                   : R${rmse_mlr:.8f}")
print()
print(f"  📐 Equation: Revenue = {mlr.intercept_:.4f}")
print(f"                       + {mlr.coef_[0]:.4f} × Price")
print(f"                       + {mlr.coef_[1]:.4f} × FreightValue")
print()
print("  Interpretation:")
print(f"  • Adding freight_value as a predictor increases explained variance")
print(f"  • β₀ ≈ 0 confirms the near-identity: revenue ≈ price + freight,")
print(f"    with coefficients β₁ ≈ 1 and β₂ ≈ 1 as expected.")
```


## 3.5 — Forecasting Method 1: 3-Month Rolling Moving Average  `P5` `P6`

The Moving Average smooths short-term fluctuations by averaging recent observations. With window k=3, the forecast for month t+1 = average of months t, t−1, and t−2.


```python
main["order_purchase_timestamp"] = pd.to_datetime(
    main["order_purchase_timestamp"]
)

main["month"] = main[
    "order_purchase_timestamp"
].dt.to_period("M")

monthly = (
    main.groupby("month")["revenue"]
    .sum()
)
```

```python
main["order_purchase_timestamp"] = pd.to_datetime(
    main["order_purchase_timestamp"]
)

main["month"] = main[
    "order_purchase_timestamp"
].dt.to_period("M")

monthly = (
    main.groupby("month")["revenue"]
    .sum()
)
```


## 3.7 — Forecasting Method 2: Simple Exponential Smoothing (SES)  `P5` `P6` `M3`

Exponential Smoothing gives **exponentially decreasing weight** to older observations. The smoothing parameter α (0 < α < 1) controls how quickly the model adapts: α close to 1 → highly reactive; α close to 0 → slow/smooth. This makes it superior to Moving Average when recent data should be weighted more heavily.

**Formula:** F(t+1) = α × Revenue(t) + (1 − α) × F(t)

## 3.9 — Critical Evaluation: Predictive Analytics for Forecasting  <span style="background:#7c3aed;color:white;padding:2px 10px;border-radius:12px;font-size:12px;font-weight:600;">D2</span>

The predictive analyses conducted in this examination generate directionally credible insights, but must be evaluated critically in terms of data quality constraints, model assumptions, and responsible business deployment.

**Regarding the Linear Regression Model:**  
The R² of 0.9943 for Simple Linear Regression is superficially impressive but analytically misleading. Since `revenue` is *definitionally* equal to `price + freight_value`, and since freight weakly correlates with price (r = 0.414), the regression largely recovers a mathematical identity rather than discovering a novel predictive relationship. This constitutes a form of circularity: the model is fitting an equation where the response variable is a direct function of the predictor. A genuinely predictive regression for business use would deploy independent exogenous predictors — such as product category, weight, customer region, or seller rating — to forecast *demand volume* or *return probability*, rather than confirming a definitional relationship between price components (James et al., 2021). The Multiple Linear Regression confirms this: R² ≈ 1.0 when both price and freight are included, because revenue is exactly their sum. These models, while technically correct, offer limited commercial predictive value.

**Regarding the Moving Average:**  
The MA-3 forecast produces intuitively reasonable outputs but carries three significant limitations: (1) It assigns equal weight to all three historical periods, meaning a revenue spike two months ago has the same influence as last month's figure — inappropriate in a fast-growing market. (2) It cannot detect or model seasonality — the November Black Friday peak observed in the 2017 data would require dedicated seasonal decomposition (e.g., Holt-Winters or STL decomposition) for proper forecasting. (3) The window size k=3 was selected pragmatically rather than through formal optimisation; a different window would produce materially different forecast values (Provost & Fawcett, 2013).

**Regarding Exponential Smoothing:**  
SES with α=0.4 shows better responsiveness to recent trends than MA-3, but shares MA's inability to model seasonality and trend simultaneously. For a platform with clear seasonal demand (Black Friday spikes, post-holiday dips), a Holt-Winters triple exponential smoothing model — which explicitly decomposes time series into level, trend, and seasonal components — would be significantly more appropriate for production forecasting.

**Data Quality Constraints:**  
The 2,454 missing delivery dates may represent cancelled orders whose revenue should be excluded from the time series. Including these records could inflate monthly revenue figures, leading to over-optimistic forecasts. Similarly, the 1,603 rows with missing product category data were imputed with the mode — which could misclassify premium-category products as standard ones, potentially distorting category-level revenue analysis.

**Responsible Business Use:**  
Organisations deploying these forecasts must understand their inherent uncertainty. A business committing to inventory orders or logistics contracts based solely on a 3-month moving average is exposed to significant demand forecasting error. Forecasts should always be presented with confidence intervals, model performance metrics (MAE, RMSE), and explicit assumptions stated. As Marr (2017) observes, the greatest organisational risk in analytics is not predicting incorrectly — it is communicating predictions as certainties to decision-makers who then act on them without understanding the underlying uncertainty bounds.
